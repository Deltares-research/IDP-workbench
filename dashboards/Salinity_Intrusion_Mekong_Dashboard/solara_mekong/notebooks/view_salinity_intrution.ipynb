## 1) Kernel path + Solara theme

Forces **Solara light theme** (`solara.lab.theme.dark = False`) so the dashboard stays readable with a dark Cursor/VS Code theme.


```python
import sys
from pathlib import Path

import solara

# Repo root = parent of notebooks/
REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == "notebooks":
    REPO_ROOT = REPO_ROOT.parent

if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

# Force light Solara UI even when Cursor/VS Code uses a dark theme
solara.lab.theme.dark = False

print("REPO_ROOT:", REPO_ROOT)
print("Solara dark theme:", solara.lab.theme.dark)
```

## 2) Optional: quick GeoServer ping

```python
import requests

remote = (
    "https://international-delta-platform.avi.directory.intra/geoserver"
)
try:
    r = requests.get(f"{remote}/web/", timeout=15, verify=False)
    print(f"Remote GeoServer UI: {r.status_code} ({remote}/web/)")
except Exception as exc:
    print(
        "Remote GeoServer not reachable (VPN may be required).\n"
        f"Error: {exc}"
    )
```
